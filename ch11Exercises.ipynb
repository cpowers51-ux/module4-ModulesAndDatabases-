```python
#16.4

import sqlite3

conn = sqlite3.connect("books.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS books (
title TEXT,
author TEXT,
year INTEGER
)
""")

conn.commit()

print("Database and table created successfully")
```

```python
%%writefile books2.csv
title,author,year
The Weirdstone of Brisingamen,Alan Garner,1960
Perdido Street Station, China Mieville,2000
Thud!,Terry Pratchett,2005
The Spellman Files,Lisa Lutz,2007
Small Gods, Terry Pratchett,1992
```

```python
#16.5

import csv

with open("books2.csv", "r", newline="") as file:
    reader = csv.DictReader(line for line in file if line.strip())

    for row in reader:
        cursor.execute(
            "INSERT INTO books (title, author, year) VALUES (?, ?, ?)",
            (row["title"], row["author"], int(row["year"]))
        )

conn.commit()

print("Books inserted successfully!!!!!!!")
```

```python
#16.8

import sqlalchemy as sa

engine = sa.create_engine("sqlite:///books.db")

with engine.connect() as conn:
    result = conn.execute(
        sa.text("SELECT title FROM books ORDER BY title")
    )

    for row in result:
        print(row[0])
```
